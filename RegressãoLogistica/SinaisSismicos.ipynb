```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('dados_sismicos.csv')

#valor de y, label
evento = df['Evento']

#valor de x, feature
ampFr = df[['Amplitude', 'Frequencia']]
```

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(ampFr, evento, test_size = 0.2, random_state = 42)

print(X_treino.shape)

model = LogisticRegression()

model.fit(X_treino, y_treino)

previsaoTeste = model.predict(X_teste)
previsaoSelecionado = model.predict([[6.5, 20.0]])
porcentagemPrevisaoSelecionado = model.predict_proba([[6.5, 20.0]])

resultado = np.where((previsaoTeste == 1), "tremor", "ruido")

print(resultado)
print(previsaoSelecionado)
print(porcentagemPrevisaoSelecionado)
```
