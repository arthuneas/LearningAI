Enquanto na regressão linear temos como saída um valor numérico, como, com base na sua idade, um salário (valor numérico) é um output do modelo. A regressão logistica garante uma probabilidade como saída. Esse modelo tem uma sigmoide, uma curva gerada a partir da função de exponencial e quadrática, como modelação da saída da função. 

Dessa maneira, quando o valor tende ao infinito, ele vai para um, para menos infinito, ele vai para 0, obviamente, representa-se a saída como um resultado da função (y). Por exemplo, usamos a Regressão Logística para determinar a probabilidade de um modelo ser spam ou não. Esse modelo também podes servir como classificador ou um auxiliar ao classificador. Além de ser utilizado como probabilidade, a partir dela, pode fazer-se um classificador binário: x% determina se um valor é spam ou não spam.

para o calculo da probabilidade da sigmoide, precisa-se da estrutura linear aplicado ao expoente de 'e'. Ou seja, fazemos o processo da função da regressão linear para calcular a saída da regressão logística.

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
#gera 50 campos com valor entre 0 e 10
#esse é o X
horasEstudo = np.random.randint(0, 10, 50)

#cria uma matriz onde se algum valor > 5, gera 1, caso contrário, 0.
#esse é o Y
aprovacao = np.where(horasEstudo > 5, "Aprovado", "Desaprovado") #condição, true, false

df_sint = pd.DataFrame({'horasEstudo': horasEstudo, 'aprovacao': aprovacao})
df_sint.to_csv('dadosEstudos.csv', index = False)

print("O arquivo dadosEstudos.csv foi produzido.")
```

```python
df = pd.read_csv('dadosEstudos.csv')

df.info()

display(df.sort_values(by = 'horasEstudo'))

horas = df[['horasEstudo']]
apr = df['aprovacao']
```

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(horas, apr, test_size = 0.2, random_state = 42)

print(X_treino.shape)

model = LogisticRegression().fit(X_treino, y_treino)

predictionTeste = model.predict(X_teste)
predictionSix = model.predict([[6]])
predictionThree = model.predict([[3]])

print(predictionTeste)
print(predictionSix)
print(predictionThree)
```
